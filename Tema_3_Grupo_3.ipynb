# Taller Grupo 3: Análisis de estudiantes en riesgo académico 
Este análisis nos permitirá identificar qué estudiantes están en riesgo de perder materias, en base a sus notas en Matemáticas, Física, Programación y su porcentaje de asistencia.

```python
import pandas as pd

# Creamos datos simulados de ejemplo
datos = {
    'Estudiante': ['Cindy', 'Alex', 'Kevin', 'Mauricio', 'Aracelis'],
    'Nota_Matemáticas': [6.5, 4.0, 7.2, 5.0, 3.8],
    'Nota_Física': [7.0, 3.5, 6.8, 4.8, 5.5],
    'Nota_Programación': [6.0, 4.2, 7.5, 4.9, 3.5],
    'Asistencia_%': [90, 75, 85, 60, 58]
}

df = pd.DataFrame(datos)
df
```

### Cálculo del promedio de notas
Se agregará una columna nueva con el promedio de las tres materias.

```python
df['Promedio'] = df[['Nota_Matemáticas', 'Nota_Física', 'Nota_Programación']].mean(axis=1)
df
```

### Evaluación del riesgo
Un estudiante estará en riesgo si:
- Tiene un promedio menor a 6.0 o
- Tiene una asistencia menor al 70%

```python
df['Riesgo'] = ((df['Promedio'] < 6.0) | (df['Asistencia_%'] < 70)).map({True: 'Sí', False: 'No'})
df
```

### Estudiantes en riesgo
A continuación se muestran únicamente los estudiantes que están en riesgo según los criterios definidos.

```python
estudiantes_en_riesgo = df[df['Riesgo'] == 'Sí']
estudiantes_en_riesgo
```

## Resumen de resultados

- Total de estudiantes: 5
- Estudiantes en riesgo: 3

La identificación de estudiantes en riesgo permite a la universidad tomar decisiones preventivas para mejorar su rendimiento académico y retención.
